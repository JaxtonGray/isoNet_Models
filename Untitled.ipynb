```python
from glob import glob
import os, shutil, zipfile
```

```python
models = glob(r'Models/Global*/**/*.keras', recursive=True)
```

```python
test = models[0]
model_type = test.split('/')[1]
os.path.basename(test).split('_')[1].split('.')[0]
```

```python
def rename_global_model(modelDir):
    model_type = modelDir.split('/')[1]
    runNum = os.path.basename(modelDir).split('_')[1].split('.')[0]
    return f'Model_{model_type}_{runNum}.keras'
```

```python
rename_global_model(test)
```

```python
test = r'Models/Global_B/Model_Run1/Model_Run1.keras'
shutil.copyfile(test, 'global_test.keras')
```

```python
os.getcwd()
```

```python
for m in models:
    path = os.path.dirname(m)
    newPath = os.path.join(path, rename_global_model(m))
    shutil.copy(m, newPath)
```
