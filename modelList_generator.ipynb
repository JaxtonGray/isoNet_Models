ModelList Generator
-----------------------------------------------------------------------------

```python
# Import libraries
import itertools
from glob import glob
```

## From Scratch

```python
# Define function that takes in the feature and spatial scheme lists and outputs
# a list of all possible model combinations of varying amounts of features
def generate_model_combinations(features, spatialSchemes):
    # First generate all possible feature combinations of varying amounts of features.
    # This will not include the base set ('B') as it will be added to each after
    other_features = [item for item in features if item != 'B']
    allFeatures = []

    # Generate combinations of other features (0 to all other features)
    for r in range(0, len(other_features) + 1):
        for combo in itertools.combinations(other_features, r):
            # Add 'B' to each combination
            allFeatures.append(('B',) + combo)
    # Finally combine all feature combinations with all spatial schemes
    return list(itertools.product(allFeatures, spatialSchemes))
```

```python
# Create function that will take the output of a list containing the features and schemese and output a list of model names
def generate_model_name(combined_list):
    # This will take the output generated by the generate_model_combinations which gives a tupe containing a tuple of the features, and scheme
    # ex. (('B', 'K') 'PreWinds')

    model_names = []
    # Cycle through list and combine to create singular string with the model name
    for model in combined_list:
        # Grab the joined features and scheme
        feats = ('').join(model[0])
        scheme = model[1]
        # Add to the model_names following the given structure
        model_names.append(f"{scheme}_{feats}")
    return model_names
```

```python
# Set list of features to use in models and schemes
features = ['K', 'T', 'P']
schemes = ['PreWinds', 'Global']
runs = range(1, 11)  # Runs 1 to 10
```

```python
combList = generate_model_combinations(features, schemes)
model_combinations = generate_model_name(combList)
model_combinations
```

```python
# Open a file to write the model list
with open('modelList.txt', 'w') as f:
    # Cycle through the model combinations 
    for model in model_combinations:
        # Cycle through the runs
        for run in runs:
            f.write(f"{run} {model}\n")
```

## From Failed Training Models

```python
# Grab all the slurm files available
slurmFiles = glob('SLURM_Output/model_training*')
```

```python
# Cycle through each file and find which ones timed out
failedTraining = []
for file in slurmFiles:
    with open(file, 'r') as f:
        lines = f.readlines()
        if 'Model Finished Training' not in lines[-1]:
            failedTraining.append(lines[0])
        
```

```python
reTrain = []
for file in failedTraining:
    split = file.split()
    reTrain.append((split[3], split[1]))
reTrain
```

```python
# Write the models to retrain to modelList.txt
with open("modelList.txt", 'w') as f:
    for model in reTrain:
        f.write(f"{model[0]} {model[1]}\n")
```
