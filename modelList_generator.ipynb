ModelList Generator
-----------------------------------------------------------------------------

```python
# Import libraries
import itertools
from glob import glob
```

## From Scratch

```python
# Set list of features to use in models and schemes
features = ['B']
schemes = ['Global', 'PreWinds', 'HighLowLat']
runs = range(1, 11)  # Runs 1 to 10
```

```python
# Combine each scheme with each feature in a single string with a '_' separator
model_combinations = []

for scheme, feature in itertools.product(schemes, features):
    model_combinations.append(f"{scheme}_{feature}")

model_combinations
```

```python
# Open a file to write the model list
with open('modelList.txt', 'w') as f:
    # Cycle through the model combinations 
    for model in model_combinations:
        # Cycle through the runs
        for run in runs:
            f.write(f"{run} {model}\n")
```

## From Timed Out Models

```python
# Grab all the slurm files available
slurmFiles = glob('SLURM_Output/model_training*')
```

```python
# Cycle through each file and find which ones timed out
failedTraining = []
for file in slurmFiles:
    with open(file, 'r') as f:
        lines = f.readlines()
        if 'CANCELLED' in lines[-1]:
            failedTraining.append(lines[0])
        
```

```python
reTrain = []
for file in failedTraining:
    split = file.split()
    reTrain.append((split[3], split[1]))
reTrain
```

```python
# Write the models to retrain to modelList.txt
with open("modelList.txt", 'w') as f:
    for model in reTrain:
        f.write(f"{model[0]} {model[1]}")
```
